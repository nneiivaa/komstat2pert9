## 1. Introduction to NumPy

### 1.1 What Is NumPy?
NumPy (Numerical Python) is a Python library designed for numerical and scientific computation. NumPy makes it convenient to organize numerical data, perform operations on entire arrays, and prepare data for statistical analysis, scientific modeling, and machine learning.

### 1.2 Importing NumPy
The conventional alias for NumPy is **np**.

### 1.3 Python Lists vs. NumPy Arrays
**Python** List dapat menyimpan berbagai tipe data.

**NumPy** Array menyimpan elemen dengan tipe data yang sama yang lebih efisien untuk perhitungan numerik.

Misal operator + pada List untuk menggabungkan elemen, pada NumPy Array untuk penjumlahann elemen yang bersesuaian.

```python
python_list = [2, 4, 6]
numpy_array = np.array([2, 4, 6])

print("Python list:", python_list)
print("NumPy array:", numpy_array)
print("Python type:", type(python_list))
print("NumPy type:", type(numpy_array))
```

```python
print(python_list + python_list)
print(numpy_array + numpy_array)
```

## 2. Creating NumPy Arrays
## 2.1 Creating Arrays from Python Lists
Fungsi np.array() digunakan untuk mengubah data Python menjadi NumPy Array. Jumlah dimensi ditentukan oleh struktur data yang dimasukkan.

- **Array 1D:** Berbentuk vektor.
- **Array 2D:** Berbentuk matriks (baris dan kolom).
- **Array 3D:** Kumpulan matriks 2D.
- **Array 4D:** Kumpulan array 3D.
  
Harus memiliki bentuk (shape) yang konsisten. **Shape:** Ukuran NumPy Array yang menunjukkan jumlah elemen pada setiap dimensi.

### 2.2 Creating a One-Dimensional Array (1D)
A 1D array is useful for representing measurements of a **single variable**, such as the scores of five students.

```python
student_scores = [75, 80, 92, 86, 88]
scores_1d = np.array(student_scores)

print(scores_1d)
print("Dimensions:", scores_1d.ndim)
print("Shape:", scores_1d.shape)
```

### 2.3 Creating a Two-Dimensional Array (2D)
A 2D array represents data **in rows and columns**. In the example below, each row corresponds to a student, while each column corresponds to a different assessment.

```python
student_assessments = [[80, 85, 90],
                       [75, 88, 84],
                       [92, 90, 95]]

scores_2d = np.array(student_assessments)

print(scores_2d)
print("Dimensions:", scores_2d.ndim)
print("Shape:", scores_2d.shape)
```

The shape (3, 3) indicates **three rows and three columns**.

### 2.4 Creating a Three-Dimensional Array (3D)
A 3D array can be understood as **a stack of matrices**.

```python
week_1 = np.array([[10, 12],
                   [14, 16]])
week_2 = np.array([[11, 13],
                   [15, 17]])

observations_3d = np.array([week_1, week_2])

print(observations_3d)
print("Dimensions:", observations_3d.ndim)
print("Shape:", observations_3d.shape)
```

The shape (2, 2, 2) indicates **two matrices**, each containing **two rows and two columns**. The first axis identifies which matrix is selected.

### 2.5 Creating a Four-Dimensional Array (4D)
A 4D array adds another organizational level. You can think of it as a collection of 3D arrays. Such structures appear in scientific data and batches of multidimensional observations.

```python
first_cube = np.array([[[1, 2], [3, 4]],
                       [[5, 6], [7, 8]]])
second_cube = np.array([[[11, 12], [13, 14]],
                        [[15, 16], [17, 18]]])

observations_4d = np.array([first_cube, second_cube])

print(observations_4d)
print("Dimensions:", observations_4d.ndim)
print("Shape:", observations_4d.shape)
```

### 2.6 Creating Arrays from Scratch: np.zeros()
np.zeros() directly creates an array filled with zeros.

```python
zero_vector = np.zeros(5)
zero_matrix = np.zeros((3, 4))

print("1D zeros:")
print(zero_vector)
print("2D zeros:")
print(zero_matrix)
```

### 2.7 Creating Random Arrays: np.random.random()
np.random.random() menghasilkan bilangan desimal acak antara **0 hingga kurang dari 1**. Penggunaan seed membuat hasil bilangan acak tetap sama setiap kali kode dijalankan.

```python
np.random.seed(42)
random_measurements = np.random.random((2, 3))

print(random_measurements)
print("Shape:", random_measurements.shape)
```

### 2.8 Creating Sequences: np.arange()
np.arange(start, stop, step) digunakan untuk membuat deret angka dengan nilai awal (start), batas akhir (stop), dan jarak antarangka (step). Nilai stop tidak disertakan. Secara default, start = 0 dan step = 1.

```python
sequence_a = np.arange(5)
sequence_b = np.arange(2, 11, 2)
sequence_c = np.arange(-3, 4, 3)

print("Default sequence:", sequence_a)
print("Even-number sequence:", sequence_b)
print("Stepped sequence:", sequence_c)
```

## 3. Array Attributes and Data Types
### 3.1 Array Dimensionality and Axes
Axis adalah sumbu yang menunjukkan arah dimensi pada NumPy Array. Pada array 2D, axis=0 menunjukkan arah baris dan axis=1 menunjukkan arah kolom. Pada array 3D, terdapat tiga sumbu, yaitu lapisan, baris, dan kolom.

- **ndim:** Jumlah dimensi atau sumbu.
- **shape:** Ukuran setiap dimensi.
- **size:** Jumlah seluruh elemen array.

```python
measurement_matrix = np.array([[12, 15, 18, 21],
                               [14, 16, 20, 22],
                               [11, 13, 17, 19]])

print("Array:")
print(measurement_matrix)
print("ndim:", measurement_matrix.ndim)
print("shape:", measurement_matrix.shape)
print("size:", measurement_matrix.size)
print("dtype:", measurement_matrix.dtype)
```

### 3.2 Understanding Shape and Size
A shape of (3, 4) means the array has three rows and four columns; its size is therefore 3 × 4 = 12 elements. For a 3D or 4D array, the total size is the product of all entries in its shape.

```python
print("1D:", scores_1d.ndim, scores_1d.shape, scores_1d.size)
print("2D:", scores_2d.ndim, scores_2d.shape, scores_2d.size)
print("3D:", observations_3d.ndim, observations_3d.shape, observations_3d.size)
print("4D:", observations_4d.ndim, observations_4d.shape, observations_4d.size)
```

### 3.3 Common NumPy Data Types (dtype)
NumPy includes types for integers (int), floating-point numbers (float), Boolean values (bool), and strings (str). The actual dtype shown by NumPy provides more detail, such as the number of bits used for a numeric value.

Because NumPy arrays use a common data type, combining different kinds of values may cause automatic type conversion.

```python
integer_values = np.array([1, 2, 3])
float_values = np.array([1.5, 2.5, 3.5])
boolean_values = np.array([True, False, True])
string_values = np.array(["low", "medium", "high"])

print("Integer:", integer_values.dtype)
print("Float:", float_values.dtype)
print("Boolean:", boolean_values.dtype)
print("String:", string_values.dtype)
```

### 3.4 Specifying a Data Type Explicitly
Use the dtype argument in np.array() to request a specific type. A smaller numeric type may occupy less memory, but it also has a more limited range or precision.

```python
measurements_float32 = np.array([1.25, 2.75, 3.50], dtype=np.float32)
counts_int32 = np.array([10, 20, 30], dtype=np.int32)

print(measurements_float32, measurements_float32.dtype)
print(counts_int32, counts_int32.dtype)
```

```python
measurements_int32 = np.array([1.25, 2.75, 3.50], dtype=np.int32)
measurements_int32 
```

### 3.5 Type Conversion with astype()
The .astype() method returns an array converted to the specified data type. A common application is converting strings that contain valid numbers into numeric values.

Be careful: converting a floating-point number to an integer discards its fractional part; it is not the same as rounding to the nearest integer.

```python
text_scores = np.array(["70", "85", "90"])
numeric_scores = text_scores.astype(np.int64)

print("Before:", text_scores, text_scores.dtype)
print("After:", numeric_scores, numeric_scores.dtype)
```

```python
continuous_values = np.array([1.9, 2.1, 3.8])
integer_conversion = continuous_values.astype(int)

print("Original:", continuous_values)
print("Converted:", integer_conversion)
```

### 3.6 Type Coercion
Type coercion occurs when NumPy automatically chooses a common data type for the elements used to create an array. Examples include converting integers to floats when a decimal is included, or converting numeric values to strings when they are mixed with text.

**Always inspect .dtype before numerical analysis.**

```python
coercion_a = np.array([10, 20.5, 30])
coercion_b = np.array([True, False, 5])
coercion_c = np.array([1, "two", 3])

print(coercion_a, coercion_a.dtype)
print(coercion_b, coercion_b.dtype)
print(coercion_c, coercion_c.dtype)
```

## 4. Reshaping Arrays
### 4.1 Understanding Array Shape
The **shape** describes how the elements are organized. Importantly, a 1D array with six elements has shape (6,), not (1, 6). A row-shaped 2D array and a column-shaped 2D array have different shapes, even when they contain the same six values.

```python
base_values = np.arange(1, 7)

row_array = base_values.reshape(1, 6)
column_array = base_values.reshape(6, 1)

print("1D shape:", base_values.shape)
print("Row shape:", row_array.shape)
print(row_array)
print("Column shape:", column_array.shape)
print(column_array)
```

### 4.2 Reshaping Arrays with reshape()
.reshape() changes the arrangement of the elements without changing the total number of elements. A 1D array containing 12 values can be reshaped into (3, 4), (4, 3), or (2, 2, 3) because each shape contains exactly 12 positions.

For this basic example, NumPy reads the elements in their existing order and organizes them into the new shape.

```python
values_12 = np.arange(1, 13)

matrix_3x4 = values_12.reshape(3, 4)
matrix_4x3 = values_12.reshape(4, 3)
array_2x2x3 = values_12.reshape(2, 2, 3)

print("Original:", values_12)
print("3 × 4:")
print(matrix_3x4)
print("4 × 3:")
print(matrix_4x3)
print("2 × 2 × 3:")
print(array_2x2x3)
```

### 4.3 Reshape Compatibility
The total number of elements **must match** the requested shape. For example, an array of size 12 cannot be reshaped into (5, 3), because that shape requires 15 elements.

NumPy can calculate **one** missing dimension when you specify -1.

```python
reshaped_automatically = values_12.reshape(3, -1)

print(reshaped_automatically)
print("Inferred shape:", reshaped_automatically.shape)
```

### 4.4 Flattening Arrays with flatten()
.flatten() converts a multidimensional array into a **new 1D copy**. It is useful when the analysis requires all elements in a single sequence.

```python
original_matrix = np.array([[1, 2],
                            [5, 7],
                            [6, 6]])
flattened_values = original_matrix.flatten()

print("Original:")
print(original_matrix)
print("Flattened:")
print(flattened_values)
print("Flattened shape:", flattened_values.shape)
```

## 5. Basic Array Indexing and Slicing
###  5.1 Indexing a One-Dimensional Array (1D)
**Indexing** selects an element based on its position. Python uses **zero-based indexing**, so index 0 is the first element, index 1 is the second, and so on.

Negative indices count backward from the end: -1 is the last element.

```python
values = np.array([2, 4, 6, 8, 10])

print("First element:", values[0])
print("Fourth element:", values[3])
print("Last element:", values[-1])
```

### 5.2 Indexing a Two-Dimensional Array (2D)
A 2D array requires a **row index** and a **column index** to select one element. The syntax is array[row, column]. Both indices start from zero.

```python
marks = np.array([[80, 85, 90],
                  [75, 88, 84],
                  [92, 90, 95]])

print(marks)
print("Row 1, column 2:", marks[1, 2])
print("Row 2, column 0:", marks[2, 0])
```

### 5.3 Selecting Rows and Columns
Use a single index to select a row (array[row]) or combine an index with : to select a complete column (array[:, column]). The colon means **all positions** along that axis.

```python
print("First row:", marks[0])
print("Second row:", marks[1, :])
print("Third column:", marks[:, 2])
print("First column:", marks[:, 0])
```

### 5.4 Slicing a One-Dimensional Array (1D)
**Slicing** retrieves several consecutive elements. The general syntax is array[start:stop].

- start is included.
- stop is excluded.
- An omitted start means the beginning of the array.
- An omitted stop means the end of the array.

```python
sequence = np.array([10, 20, 30, 40, 50, 60])

print("Positions 1 to 3:", sequence[1:4])
print("First three:", sequence[:3])
print("From position 3 onward:", sequence[3:])
print("All elements:", sequence[:])
```

### 5.5 Slicing a Two-Dimensional Array (2D)
A 2D slice specifies a range of rows and a range of columns: array[row_start:row_stop, col_start:col_stop]. This is useful for extracting a submatrix.

```python
number_grid = np.arange(1, 26).reshape(5, 5)

print("Original matrix:")
print(number_grid)
print("Rows 1–3, columns 2–4:")
print(number_grid[1:4, 2:5])
```

### 5.6 Slicing with Steps
A slice can include a third argument: array[start:stop:step]. The step specifies the interval between selected positions.

- ::2 selects every second element, starting at the beginning.
- 1::2 selects every second element, starting at index 1.
- [::-1] reverses a 1D array.

Steps can also be applied separately to the row and column axes of a 2D array.

```python
print("Every second element:", sequence[::2])
print("Every second from index 1:", sequence[1::2])
print("Reversed:", sequence[::-1])
print("Every second row and column:")
print(number_grid[::2, ::2])
```
